# 🔬 Pengujian Pipeline LBPH (Local Binary Pattern Histogram)

Notebook ini memvisualisasikan seluruh tahapan pipeline pengenalan wajah berbasis LBP:
1. **Deteksi Wajah (ROI)** menggunakan YuNet
2. **Crop Wajah**
3. **Grayscale**
4. **Histogram Equalization** (`cv2.equalizeHist`)
5. **Pola Tekstur LBP** (2D Map)
6. **Vektor Histogram Spasial LBP** (Grid 8x8 x 256 bin)
7. **Pencocokan Jarak Chi-Square & Cosine Similarity** terhadap database `api_data/faces_lbph.json`

```python
# Cell 1: Setup Library & Load Komponen Codebase
import json
import glob
from pathlib import Path
import cv2
import numpy as np
import matplotlib.pyplot as plt

from core.detector import Detector
from core.engine import _extract_lbp, _resize_for_detection

# Inisialisasi detector YuNet & database prototype LBPH
detector = Detector("face_detection_yunet_2023mar.onnx", min_score=0.5, strict_geometry=False)
DB_PATH = Path("api_data/faces_lbph.json")
lbph_db = json.loads(DB_PATH.read_text(encoding="utf-8")) if DB_PATH.exists() else {}

print(f"Detektor aktif: {detector.kind}")
print(f"Database LBPH terload: {list(lbph_db.keys())}")
```

```python
# Cell 2: Fungsi Pengujian Pipeline dengan Visualisasi Subplot

def uji_lbph(image_path, threshold_chi2=34.0, sim_threshold=0.45):
    """
    Menjalankan pipeline pengenalan wajah LBPH dan memvisualisasikan
    setiap tahap pemrosesan dalam bentuk subplot matplotlib.
    """
    frame = cv2.imread(str(image_path))
    if frame is None:
        print(f"❌ File tidak ditemukan: {image_path}")
        return None

    frame = _resize_for_detection(frame, max_dim=1280)
    details = detector.detect_detailed(frame)
    accepted = [d for d in details if d.get("accepted") or d.get("score", 0) >= 0.5]
    if not accepted:
        print(f"❌ Tidak ada wajah terdeteksi di: {image_path}")
        return None

    d = accepted[0]
    bx, by, bw, bh = int(d["x"]), int(d["y"]), int(d["w"]), int(d["h"])
    h_img, w_img = frame.shape[:2]
    bx, by = max(0, bx), max(0, by)
    bw, bh = min(w_img - bx, bw), min(h_img - by, bh)

    # --- 1. Pipeline Ekstraksi Fitur LBP ---
    face_crop = frame[by:by+bh, bx:bx+bw]
    gray = cv2.cvtColor(face_crop, cv2.COLOR_BGR2GRAY)
    equalized = cv2.equalizeHist(gray)
    vec, lbp_img = _extract_lbp(equalized, return_image=True)

    # --- 2. Pencocokan dengan Database faces_lbph.json ---
    best_name, best_dist, best_sim = "Unknown", 999.0, 0.0
    distances = {}
    vec_norm = vec / (np.linalg.norm(vec) + 1e-7)

    for name, data in lbph_db.items():
        proto = np.asarray(data["histogram"], dtype=np.float32)
        proto_norm = proto / (np.linalg.norm(proto) + 1e-7)
        cos_sim = float(np.dot(vec_norm, proto_norm))
        chi2 = float(0.5 * np.sum(((vec - proto) ** 2) / (vec + proto + 1e-7)))
        distances[name] = (chi2, cos_sim)
        if cos_sim > best_sim:
            best_sim, best_name, best_dist = cos_sim, name, chi2

    is_match = bool(best_sim >= sim_threshold and best_dist <= threshold_chi2)
    pred_label = best_name if is_match else "Unknown"

    # --- 3. Subplot Visualisasi Proses ---
    fig, axes = plt.subplots(2, 3, figsize=(15, 9))
    title_color = "green" if is_match else "red"
    fig.suptitle(f"Prediksi: {pred_label} | Chi2: {best_dist:.2f} (Max: {threshold_chi2}) | Kesesuaian: {best_sim*100:.1f}%",
                 fontsize=14, fontweight="bold", color=title_color)

    # Subplot 1: Deteksi Wajah (ROI)
    vis = frame.copy()
    cv2.rectangle(vis, (bx, by), (bx + bw, by + bh), (0, 255, 0), 3)
    cv2.putText(vis, pred_label, (bx, max(25, by - 10)),
                cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 0), 2)
    axes[0, 0].imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))
    axes[0, 0].set_title("1. Deteksi Wajah (ROI)")
    axes[0, 0].axis("off")

    # Subplot 2: Crop Wajah Asli
    axes[0, 1].imshow(cv2.cvtColor(face_crop, cv2.COLOR_BGR2RGB))
    axes[0, 1].set_title(f"2. Crop Wajah ({bw}x{bh})")
    axes[0, 1].axis("off")

    # Subplot 3: Grayscale
    axes[0, 2].imshow(gray, cmap="gray")
    axes[0, 2].set_title("3. Grayscale")
    axes[0, 2].axis("off")

    # Subplot 4: Histogram Equalization
    axes[1, 0].imshow(equalized, cmap="gray")
    axes[1, 0].set_title("4. Equalize Histogram")
    axes[1, 0].axis("off")

    # Subplot 5: Pola Tekstur LBP (2D Map)
    axes[1, 1].imshow(lbp_img, cmap="gray")
    axes[1, 1].set_title("5. Pola Tekstur LBP")
    axes[1, 1].axis("off")

    # Subplot 6: Vektor Histogram LBP (1D Plot)
    axes[1, 2].plot(vec, color="steelblue", linewidth=0.7)
    axes[1, 2].set_title(f"6. Vektor Histogram ({len(vec)} D)")
    axes[1, 2].set_xlabel("Bin Fitur")
    axes[1, 2].set_ylabel("Frekuensi")
    axes[1, 2].grid(True, linestyle="--", alpha=0.5)

    plt.tight_layout()
    plt.show()

    # Detail Jarak Chi-Square
    print(f"File: {image_path}")
    print("Hasil Pencocokan Database:")
    for name, (c2, sim) in distances.items():
        status = "MATCH" if name == best_name and is_match else "-"
        print(f"  [{status:5s}] {name:10s} -> Chi2: {c2:6.2f} (<= {threshold_chi2}), Cosine: {sim*100:5.1f}%")
    print("-" * 60)
    return pred_label
```

```python
# Cell 3: Uji 1 Foto Sampel
# Ubah path ini ke foto apa pun yang ingin Anda uji
sample_files = glob.glob("../faza/*.jpg") + glob.glob("../raffin/*.jpg") + glob.glob("../rafy/*.jpg")

if sample_files:
    test_img = sample_files[0]
    print(f"Menguji file: {test_img}")
    uji_lbph(test_img)
else:
    print("Contoh foto tidak ditemukan. Jalankan dengan: uji_lbph('path/ke/foto.jpg')")
```

```python
# Cell 4: Evaluasi Ringkas (1 Foto per Identitas)
for person in ["faza", "raffin", "rafy"]:
    imgs = glob.glob(f"../{person}/*.jpg")
    if imgs:
        print(f"\n=== Uji Identitas: {person.upper()} ===")
        uji_lbph(imgs[0])
```
