# SFace Face Recognition & Embedding Pipeline

Pipeline minimal berbasis **OpenCV Native** (`cv2.FaceDetectorYN` + `cv2.FaceRecognizerSF`):
1. **Deteksi wajah & 5 landmark** via YuNet (`FaceDetectorYN`).
2. **Face alignment & crop (112x112)** via `FaceRecognizerSF.alignCrop`.
3. **Ekstraksi vektor embedding (128-D L2-normalized)** via `FaceRecognizerSF.feature`.
4. **Penyimpanan galeri database** ke JSON sebagai patokan pencocokan.
5. **Pencocokan wajah (Cosine similarity)** dengan threshold standar `0.363`.

```python
# 1. Inisialisasi Model YuNet (Deteksi) dan SFace (Pengenalan)
import json
from pathlib import Path
import cv2
import numpy as np

yunet_path = "face_detection_yunet_2023mar.onnx"
sface_path = "models/face_recognition_sface_2021dec.onnx"

detector = cv2.FaceDetectorYN.create(yunet_path, "", (320, 320), score_threshold=0.8)
recognizer = cv2.FaceRecognizerSF.create(sface_path, "")
print("Model YuNet dan SFace siap.")
```

```python
# 2. Ekstraksi Embedding 128-D dari 1 Gambar
def get_embedding(img_input):
    img = cv2.imread(str(img_input)) if isinstance(img_input, (str, Path)) else img_input
    if img is None:
        return None

    h, w = img.shape[:2]
    detector.setInputSize((w, h))
    _, faces = detector.detect(img)
    if faces is None or len(faces) == 0:
        return None

    # faces[0]: bounding box + 5 landmark mata, hidung, sudut mulut
    aligned = recognizer.alignCrop(img, faces[0])
    feat = recognizer.feature(aligned)[0]
    norm = np.linalg.norm(feat)
    return (feat / norm).tolist() if norm > 0 else feat.tolist()
```

```python
# 3. Bangun Database Galeri dari Folder Dataset & Simpan ke JSON
# Format folder yang dibaca: dataset/<NamaOrang>/<foto.jpg>
def build_gallery(dataset_dir="dataset", output_file="gallery_sface.json"):
    gallery = {}
    path = Path(dataset_dir)
    if not path.exists():
        print(f"Folder '{dataset_dir}' belum ada. Buat folder dan susun foto per nama.")
        return gallery

    exts = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
    for person in sorted(path.iterdir()):
        if not person.is_dir():
            continue
        embs = [get_embedding(f) for f in person.glob("*.*") if f.suffix.lower() in exts]
        embs = [e for e in embs if e is not None]
        if embs:
            mean_emb = np.mean(embs, axis=0)
            mean_emb = mean_emb / np.linalg.norm(mean_emb)
            gallery[person.name] = mean_emb.tolist()
            print(f"Registered {person.name}: {len(embs)} foto")

    with open(output_file, "w") as f:
        json.dump(gallery, f, indent=2)
    print(f"Galeri tersimpan ke '{output_file}' ({len(gallery)} orang).")
    return gallery

gallery = build_gallery()
```

```python
# 4. Verifikasi / Pencocokan Wajah Uji ke Galeri
def recognize(test_img, gallery_dict, threshold=0.363):
    query_emb = get_embedding(test_img)
    if query_emb is None:
        return "Wajah tidak terdeteksi", 0.0

    q_arr = np.array([query_emb], dtype=np.float32)
    best_name, best_score = "Unknown", -1.0

    for name, ref_emb in gallery_dict.items():
        r_arr = np.array([ref_emb], dtype=np.float32)
        score = recognizer.match(q_arr, r_arr, cv2.FaceRecognizerSF_FR_COSINE)
        if score > best_score:
            best_score = score
            if score >= threshold:
                best_name = name

    return best_name, float(best_score)
```

```python
# 5. Self-Check Demo Pipeline
mock_face = np.array([50, 50, 100, 100, 70, 70, 130, 70, 100, 100, 80, 130, 120, 130, 0.99], dtype=np.float32)
dummy_img = np.zeros((300, 300, 3), dtype=np.uint8)
aligned = recognizer.alignCrop(dummy_img, mock_face)
feat = recognizer.feature(aligned)[0]

assert len(feat) == 128, "Embedding harus 128 dimensi!"
print(f"Self-Check OK: Aligned shape={aligned.shape}, Embedding length={len(feat)}-D")
```
