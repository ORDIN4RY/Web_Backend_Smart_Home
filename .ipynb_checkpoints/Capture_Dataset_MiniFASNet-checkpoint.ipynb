# 📸 Dataset Collector untuk Fine-Tuning MiniFASNet (Anti-Spoofing)

Notebook ini digunakan untuk mengumpulkan dataset **Face Anti-Spoofing (FAS)** langsung dari webcam laptop secara interaktif.
Dataset yang terkumpul akan langsung diproses sesuai spesifikasi input **MiniFASNet** (patch konteks wajah skala 2.7x, resolusi 80×80 pixel).

---

### 🎯 Kategori Data yang Dibutuhkan untuk Fine-Tuning:
1. **REAL (Label 0 / `real`)**: Wajah orang asli di depan kamera.
   * *Variasi yang disarankan*: ekspresi (senyum/datar), orientasi kepala (lurus, tengok kiri/kanan 15-30°, menunduk/mendongak), jarak (dekat 30cm, sedang 60cm, jauh 1m), dan kondisi cahaya (lampu ruangan, backlight, remang).
2. **SPOOF (Label 1 / `spoof`)**: Serangan liveness / wajah palsu.
   * **Screen Replay Attack**: Arahkan layar HP / tablet / laptop yang menampilkan foto atau video wajah ke arah webcam.
   * **Print Attack**: Cetak foto wajah di kertas HVS / foto warna, lalu arahkan ke webcam.
   * **Replay Video**: Putar video gerakan wajah di layar HP di depan webcam.

---

### ⌨️ Kontrol Keyboard saat Kamera Terbuka:
| Tombol | Aksi |
| :--- | :--- |
| **`[R]`** | Ambil 1 foto **REAL** (Wajah Asli) |
| **`[S]`** | Ambil 1 foto **SPOOF** (Wajah Palsu / Layar HP / Kertas) |
| **`[B]`** | **Burst Capture**: Ambil 10 foto otomatis berturut-turut (cepat) |
| **`[C]`** | Ganti sub-kategori Spoof (`screen` / `paper` / `general`) |
| **`[Q]`** atau **`[ESC]`** | Selesai & Tutup Kamera |

```python
# Cell 1: Import Library & Persiapan Struktur Direktori
import os
import sys
import cv2
import time
import datetime
import csv
import glob
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Konfigurasi Folder Dataset
DATASET_ROOT = Path("dataset_fas")
DIRS = {
    "real_raw": DATASET_ROOT / "real" / "raw",
    "real_crop": DATASET_ROOT / "real" / "crop_2.7x",
    "spoof_raw": DATASET_ROOT / "spoof" / "raw",
    "spoof_crop": DATASET_ROOT / "spoof" / "crop_2.7x",
}

# Buat semua folder jika belum ada
for d in DIRS.values():
    d.mkdir(parents=True, exist_ok=True)

CSV_PATH = DATASET_ROOT / "annotations.csv"
if not CSV_PATH.exists():
    with open(CSV_PATH, mode="w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow(["timestamp", "filename", "label", "category", "sub_type", "bbox_x", "bbox_y", "bbox_w", "bbox_h"])

print("✅ Struktur folder dataset siap di:", DATASET_ROOT.resolve())
```

```python
# Cell 2: Modul Deteksi Wajah & FaceCropper (Spesifikasi MiniFASNet)

class MiniFASCropper:
    """
    Cropper khusus MiniFASNet.
    Memperluas bounding box wajah (scale 2.7x) agar mencakup area sekitar
    (tepi layar HP, batas kertas, tekstur kulit, dll.) lalu resize ke 80x80.
    """
    @staticmethod
    def get_expanded_box(src_w, src_h, bbox, scale=2.7):
        x, y, bw, bh = bbox
        if bh > 0 and bw > 0:
            scale = min((src_h - 1) / bh, min((src_w - 1) / bw, scale))
        else:
            scale = 1.0

        new_w = bw * scale
        new_h = bh * scale
        cx = x + bw / 2.0
        cy = y + bh / 2.0

        left = cx - new_w / 2.0
        top = cy - new_h / 2.0
        right = cx + new_w / 2.0
        bottom = cy + new_h / 2.0

        if left < 0:
            right -= left
            left = 0
        if top < 0:
            bottom -= top
            top = 0
        if right > src_w - 1:
            left -= (right - src_w + 1)
            right = src_w - 1
        if bottom > src_h - 1:
            top -= (bottom - src_h + 1)
            bottom = src_h - 1

        left = max(0, int(left))
        top = max(0, int(top))
        right = min(src_w - 1, int(right))
        bottom = min(src_h - 1, int(bottom))
        return left, top, right, bottom

    @classmethod
    def crop(cls, img, bbox, scale=2.7, out_size=(80, 80)):
        src_h, src_w = img.shape[:2]
        l, t, r, b = cls.get_expanded_box(src_w, src_h, bbox, scale=scale)
        patch = img[t : b + 1, l : r + 1]
        if patch.size == 0:
            return cv2.resize(img, out_size)
        return cv2.resize(patch, out_size, interpolation=cv2.INTER_LINEAR)


# Setup Detektor Wajah (Coba YuNet ONNX dulu, fallback ke Haar Cascade)
YUNET_PATH = Path("face_detection_yunet_2023mar.onnx")
detector_type = "none"
yunet_detector = None
haar_detector = None

if YUNET_PATH.exists():
    try:
        yunet_detector = cv2.FaceDetectorYN.create(str(YUNET_PATH), "", (320, 320), 0.5, 0.3, 5000)
        detector_type = "yunet"
        print("✅ Menggunakan Face Detector: YuNet ONNX (Presisi Tinggi)")
    except Exception as e:
        print(f"⚠️ YuNet gagal ({e}), beralih ke Haar Cascade.")

if detector_type == "none":
    cascade_file = cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
    haar_detector = cv2.CascadeClassifier(cascade_file)
    detector_type = "haar"
    print("✅ Menggunakan Face Detector: Haar Cascade")

def detect_primary_face(frame):
    """
    Deteksi wajah terbesar dalam frame.
    Kembalikan (x, y, w, h) atau None jika tidak ditemukan.
    """
    h, w = frame.shape[:2]
    if detector_type == "yunet":
        yunet_detector.setInputSize((w, h))
        _, faces = yunet_detector.detect(frame)
        if faces is not None and len(faces) > 0:
            # Cari yang memiliki area box terbesar
            best_face = max(faces, key=lambda f: f[2] * f[3])
            x, y, bw, bh = int(best_face[0]), int(best_face[1]), int(best_face[2]), int(best_face[3])
            # Clamp ke boundary frame
            x, y = max(0, x), max(0, y)
            bw, bh = min(w - x, bw), min(h - y, bh)
            return (x, y, bw, bh)
    else:
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        faces = haar_detector.detectMultiScale(gray, scaleFactor=1.2, minNeighbors=5, minSize=(60, 60))
        if len(faces) > 0:
            best_face = max(faces, key=lambda f: f[2] * f[3])
            return tuple(int(v) for v in best_face)
    return None
```

```python
# Cell 3: Aplikasi Interaktif Pengambil Foto Webcam

def run_capture_tool(camera_index=0):
    """
    Buka jendela OpenCV webcam untuk mengambil data training FAS.
    Tekan 'r' untuk REAL, 's' untuk SPOOF, 'b' untuk burst capture, 'c' ganti tipe spoof, 'q' keluar.
    """
    cap = cv2.VideoCapture(camera_index)
    if not cap.isOpened():
        print(f"❌ Gagal membuka kamera dengan index {camera_index}. Coba ganti camera_index=1.")
        return

    # Atur resolusi kamera standar 720p / 640x480
    cap.set(cv2.CAP_PROP_FRAME_WIDTH, 1280)
    cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 720)

    spoof_subtypes = ["screen_hp", "paper_print", "tablet_screen", "general"]
    current_sub_idx = 0

    # Hitung data yang sudah ada sebelumnya
    count_real = len(list(DIRS["real_raw"].glob("*.jpg")))
    count_spoof = len(list(DIRS["spoof_raw"].glob("*.jpg")))

    burst_mode = False
    burst_label = None
    burst_remaining = 0
    last_burst_time = 0.0

    flash_color = None
    flash_timer = 0

    window_name = "MiniFASNet Data Capture (Tekan [Q] untuk Keluar)"
    cv2.namedWindow(window_name, cv2.WINDOW_NORMAL)
    cv2.resizeWindow(window_name, 960, 600)

    print("\n🎥 Kamera aktif! Klik pada jendela '" + window_name + "' lalu gunakan keyboard:")
    print("   [R] -> Simpan Wajah ASLI (REAL)")
    print("   [S] -> Simpan Wajah PALSU (SPOOF)")
    print("   [B] -> Burst 10x Foto (pilih R atau S saat ditanya)")
    print("   [C] -> Ganti subkategori spoof")
    print("   [Q] -> Selesai\n")

    def save_sample(frame, bbox, is_real, sub_type):
        nonlocal count_real, count_spoof
        now_str = datetime.datetime.now().strftime("%Y%m%d_%H%M%S_%f")[:19]
        label_name = "real" if is_real else "spoof"
        target_cat = "real" if is_real else f"spoof_{sub_type}"
        file_base = f"{label_name}_{now_str}.jpg"

        # 1. Simpan Full Raw Frame
        raw_path = DIRS[f"{label_name}_raw"] / file_base
        cv2.imwrite(str(raw_path), frame)

        # 2. Simpan Crop Patch 2.7x (80x80)
        crop_patch = MiniFASCropper.crop(frame, bbox, scale=2.7, out_size=(80, 80))
        crop_path = DIRS[f"{label_name}_crop"] / file_base
        cv2.imwrite(str(crop_path), crop_patch)

        # 3. Log ke CSV
        with open(CSV_PATH, mode="a", newline="", encoding="utf-8") as f:
            writer = csv.writer(f)
            writer.writerow([
                datetime.datetime.now().isoformat(),
                file_base,
                0 if is_real else 1,
                label_name,
                sub_type if not is_real else "live_face",
                bbox[0], bbox[1], bbox[2], bbox[3]
            ])

        if is_real:
            count_real += 1
        else:
            count_spoof += 1

    try:
        while True:
            ret, frame = cap.read()
            if not ret:
                print("⚠️ Gagal membaca frame dari webcam.")
                break

            # Flip horizontal agar seperti cermin (mirror preview)
            frame = cv2.flip(frame, 1)
            display_frame = frame.copy()
            h, w = frame.shape[:2]

            # Deteksi wajah
            bbox = detect_primary_face(frame)

            # Gambar panduan box jika ada wajah
            if bbox is not None:
                bx, by, bw, bh = bbox
                # Box wajah utama (hijau)
                cv2.rectangle(display_frame, (bx, by), (bx + bw, by + bh), (0, 255, 0), 2)

                # Box konteks 2.7x MiniFASNet (kuning)
                l, t, r, b = MiniFASCropper.get_expanded_box(w, h, bbox, scale=2.7)
                cv2.rectangle(display_frame, (l, t), (r, b), (0, 220, 255), 1)
                cv2.putText(display_frame, "Konteks MiniFASNet (2.7x)", (l, max(15, t - 5)),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.45, (0, 220, 255), 1)
            else:
                cv2.putText(display_frame, "Wajah Tidak Terdeteksi! Pastikan wajah terlihat jelas.",
                            (30, 80), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 255), 2)

            # Auto Burst Handling
            if burst_mode and burst_remaining > 0 and bbox is not None:
                now_t = time.time()
                if now_t - last_burst_time >= 0.2:  # interval 200ms per frame
                    is_real_flag = (burst_label == "real")
                    save_sample(frame, bbox, is_real_flag, spoof_subtypes[current_sub_idx])
                    burst_remaining -= 1
                    last_burst_time = now_t
                    flash_color = (0, 255, 0) if is_real_flag else (0, 0, 255)
                    flash_timer = 3
                    if burst_remaining == 0:
                        burst_mode = False

            # Efek Flash Visual saat menyimpan foto
            if flash_timer > 0:
                cv2.rectangle(display_frame, (0, 0), (w, h), flash_color, 12)
                flash_timer -= 1

            # Overlay HUD Info Panel
            hud_bg = display_frame.copy()
            cv2.rectangle(hud_bg, (0, 0), (w, 55), (20, 20, 20), -1)
            cv2.rectangle(hud_bg, (0, h - 45), (w, h), (20, 20, 20), -1)
            display_frame = cv2.addWeighted(hud_bg, 0.65, display_frame, 0.35, 0)

            # Teks Status Atas
            cv2.putText(display_frame, f"REAL: {count_real}", (25, 36),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.85, (0, 255, 100), 2)
            cv2.putText(display_frame, f"SPOOF: {count_spoof}", (210, 36),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.85, (0, 80, 255), 2)
            cv2.putText(display_frame, f"Sub-Spoof: [{spoof_subtypes[current_sub_idx]}] (Tekan C utk ganti)",
                        (420, 35), cv2.FONT_HERSHEY_SIMPLEX, 0.55, (220, 220, 220), 1)

            # Teks Petunjuk Bawah
            if burst_mode:
                guide_txt = f"BURST RECORDING ({burst_label.upper()}) ... Sisa {burst_remaining} frame!"
                cv2.putText(display_frame, guide_txt, (25, h - 16),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.65, (0, 255, 255), 2)
            else:
                guide_txt = "[R] Simpan REAL  |  [S] Simpan SPOOF  |  [B] Burst 10x  |  [Q] Keluar"
                cv2.putText(display_frame, guide_txt, (25, h - 16),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.55, (255, 255, 255), 1)

            cv2.imshow(window_name, display_frame)
            key = cv2.waitKey(1) & 0xFF

            if key in [ord('q'), 27]:  # 'q' atau ESC
                print("🛑 Menghentikan capture data.")
                break
            elif key in [ord('r'), ord('R')]:
                if bbox is not None:
                    save_sample(frame, bbox, is_real=True, sub_type="live")
                    flash_color = (0, 255, 0)
                    flash_timer = 4
                    print(f"📸 Disimpan REAL (#{count_real})")
                else:
                    print("⚠️ Gagal: Tidak ada wajah yang terdeteksi!")
            elif key in [ord('s'), ord('S')]:
                if bbox is not None:
                    save_sample(frame, bbox, is_real=False, sub_type=spoof_subtypes[current_sub_idx])
                    flash_color = (0, 0, 255)
                    flash_timer = 4
                    print(f"🚨 Disimpan SPOOF (#{count_spoof}) - {spoof_subtypes[current_sub_idx]}")
                else:
                    print("⚠️ Gagal: Tidak ada wajah/layar yang terdeteksi!")
            elif key in [ord('b'), ord('B')]:
                # Mulai burst mode 10 frame untuk label aktif terakhir (default real)
                burst_mode = True
                burst_remaining = 10
                burst_label = "real" if count_real <= count_spoof else "spoof"
                print(f"⚡ Memulai BURST CAPTURE 10 foto untuk: {burst_label.upper()}...")
            elif key in [ord('c'), ord('C')]:
                current_sub_idx = (current_sub_idx + 1) % len(spoof_subtypes)
                print(f"🔄 Sub-kategori spoof diganti ke: {spoof_subtypes[current_sub_idx]}")

    finally:
        cap.release()
        cv2.destroyAllWindows()
        print(f"\n🎉 Selesai! Total foto tersimpan: REAL = {count_real}, SPOOF = {count_spoof}")
        print(f"📁 File tersimpan di direktori: {DATASET_ROOT.resolve()}")

# Jalankan fungsi capture jika dijalankan langsung:
# (Anda bisa mengganti camera_index=0 jika kamera laptop utama, atau 1 jika ada webcam USB)
run_capture_tool(camera_index=0)
```

```python
# Cell 4: Review Dataset & Galeri Visualisasi Hasil Capture

real_crops = sorted(list(DIRS["real_crop"].glob("*.jpg")))
spoof_crops = sorted(list(DIRS["spoof_crop"].glob("*.jpg")))

print(f"📊 RINGKASAN DATASET FAS SAAT INI:")
print(f"   - Total Foto REAL  : {len(real_crops)}")
print(f"   - Total Foto SPOOF : {len(spoof_crops)}")
print(f"   - Total Keseluruhan: {len(real_crops) + len(spoof_crops)}")

# Visualisasikan sampel 5 foto REAL dan 5 foto SPOOF (crop 80x80 patch)
fig, axes = plt.subplots(2, 5, figsize=(15, 6))

for i in range(5):
    # Row 0: Real Samples
    ax_r = axes[0, i]
    if i < len(real_crops):
        img = cv2.imread(str(real_crops[-(i+1)]))
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        ax_r.imshow(img_rgb)
        ax_r.set_title(f"Real #{i+1}", color="green", fontsize=10)
    else:
        ax_r.text(0.5, 0.5, "Belum ada", ha="center", va="center")
    ax_r.axis("off")

    # Row 1: Spoof Samples
    ax_s = axes[1, i]
    if i < len(spoof_crops):
        img = cv2.imread(str(spoof_crops[-(i+1)]))
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        ax_s.imshow(img_rgb)
        ax_s.set_title(f"Spoof #{i+1}", color="red", fontsize=10)
    else:
        ax_s.text(0.5, 0.5, "Belum ada", ha="center", va="center")
    ax_s.axis("off")

plt.suptitle("Sample Patch MiniFASNet (80x80 - Scale 2.7x)", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
# Cell 5: Ekspor Split Train / Validation (80:20) untuk Training PyTorch

import random
import shutil

def prepare_train_val_split(val_ratio=0.2):
    """
    Menata dataset ke format standar PyTorch ImageFolder:
    dataset_fas_split/
      ├── train/ (80%)
      │    ├── 0_real/
      │    └── 1_spoof/
      └── val/   (20%)
           ├── 0_real/
           └── 1_spoof/
    """
    SPLIT_ROOT = Path("dataset_fas_split")
    for split in ["train", "val"]:
        for cls_name in ["0_real", "1_spoof"]:
            (SPLIT_ROOT / split / cls_name).mkdir(parents=True, exist_ok=True)

    def split_and_copy(src_files, cls_folder):
        files = list(src_files)
        random.seed(42)
        random.shuffle(files)
        val_count = int(len(files) * val_ratio)
        val_files = files[:val_count]
        train_files = files[val_count:]

        for f in train_files:
            shutil.copy2(f, SPLIT_ROOT / "train" / cls_folder / f.name)
        for f in val_files:
            shutil.copy2(f, SPLIT_ROOT / "val" / cls_folder / f.name)
        return len(train_files), len(val_files)

    r_tr, r_val = split_and_copy(DIRS["real_crop"].glob("*.jpg"), "0_real")
    s_tr, s_val = split_and_copy(DIRS["spoof_crop"].glob("*.jpg"), "1_spoof")

    print("🎉 Split Dataset Siap di folder:", SPLIT_ROOT.resolve())
    print(f"   - TRAIN: Real = {r_tr}, Spoof = {s_tr} (Total: {r_tr + s_tr})")
    print(f"   - VAL  : Real = {r_val}, Spoof = {s_val} (Total: {r_val + s_val})")
    print("\nDataset ini langsung siap dipakai oleh DataLoader PyTorch untuk fine-tuning MiniFASNet!")

# Jalankan split jika minimal sudah ada masing-masing 10 data
if len(real_crops) >= 10 and len(spoof_crops) >= 10:
    prepare_train_val_split(val_ratio=0.2)
else:
    print("💡 Kumpulkan minimal masing-masing 10 foto Real dan 10 foto Spoof sebelum menjalankan split.")
```

## 💡 Tips Pengambilan Dataset Anti-Spoofing Berkualitas Tinggi:

1. **Rasio Seimbang (Balanced Data)**:
   * Usahakan jumlah foto **REAL** dan **SPOOF** seimbang (misal masing-masing 100 - 300 foto).
2. **Variasi Serangan Layar (Screen Attack)**:
   * Putar foto wajah di HP dengan kecerahan layar HP yang bervariasi (terang, sedang, gelap).
   * Gerakkan layar HP sedikit miring untuk menangkap efek refleksi cahaya dan garis moiré (garis piksel layar) yang menjadi ciri khas wajah palsu pada MiniFASNet.
3. **Variasi Kertas (Print Attack)**:
   * Pegang kertas sedikit melengkung atau goyangkan perlahan agar model belajar membedakan tekstur kertas vs tekstur kulit asli.
4. **Konteks Sekitar Wajah (2.7x Scale)**:
   * MiniFASNet sangat mengandalkan area tepian wajah (bingkai HP, jari tangan yang memegang HP/kertas). Kotak kuning pada layar menunjukkan area konteks 2.7x yang dipotong otomatis.
